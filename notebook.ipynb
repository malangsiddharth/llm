```python
from groq import Groq
import os
client=Groq(api_key=os.environ.get('groq-api-key'))
```

```python
def llm(prompt):
    response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )
    return response.choices[0].message.content
```

```python
question="What are todays BSE India Annoucements about JSW Infrastructure Ltd"
answer=llm(question)
#print(answer)
print(answer)
```
